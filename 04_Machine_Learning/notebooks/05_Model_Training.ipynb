```python
# Enterprise FinTech Payment Intelligence Platform
## Phase 4 – Machine Learning
### Notebook 05 – Model Training

**Objective:**
Train a baseline Logistic Regression model, a Random Forest ensemble, and an XGBoost model using the SMOTE-balanced dataset. Save all models for detailed evaluation.
```

```python
import pandas as pd
import joblib
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
import warnings

warnings.filterwarnings("ignore")
```

```python
print("Loading preprocessed datasets...")

# Load the SMOTE-balanced training data
X_train_smote = pd.read_csv("X_train_smote.csv")
y_train_smote = pd.read_csv("y_train_smote.csv").values.ravel() # .ravel() converts to 1D array

# Load the untouched, scaled testing data
X_test = pd.read_csv("X_test.csv")
y_test = pd.read_csv("y_test.csv").values.ravel()

print(f"Balanced Training Data: {X_train_smote.shape}")
print(f"Untouched Testing Data: {X_test.shape}")
print("Datasets loaded successfully.")
```

```python
print("Training Missing Values")
print(X_train_smote.isnull().sum().sum())

print()

print("Testing Missing Values")
print(X_test.isnull().sum().sum())
```

```python
### Business Summary

• Balanced training dataset loaded successfully.

• Testing dataset remains untouched.

• Preprocessed features are ready for model training.

• Fraud labels successfully separated from predictor variables.
```

```python
print("=" * 60)
print("Training Logistic Regression Model")
print("=" * 60)

log_model = LogisticRegression(
    random_state=42,
    max_iter=1000,
    n_jobs=-1,
    solver='saga' # Optimized for very large datasets
)

log_model.fit(X_train_smote, y_train_smote)
print("Logistic Regression Training Completed Successfully")
```

```python
y_pred_log = log_model.predict(X_test)
y_prob_log = log_model.predict_proba(X_test)[:, 1]

train_accuracy = log_model.score(X_train_smote, y_train_smote)
test_accuracy = log_model.score(X_test, y_test)

print(f"Training Accuracy : {train_accuracy:.4f}")
print(f"Testing Accuracy  : {test_accuracy:.4f}")
```

```python
joblib.dump(log_model, "LogisticRegression_Model.pkl")
print("Logistic Regression Model Saved Successfully")
```

```python
### Logistic Regression Summary

• Logistic Regression serves as the baseline fraud detection model.

• The model was trained using the balanced training dataset created with SMOTE.

• Predictions were generated on the untouched testing dataset to simulate real-world deployment.

• The trained model has been saved for detailed evaluation in Notebook 06.
```

```python
print("=" * 60)
print("Training Random Forest Model")
print("=" * 60)

# n_estimators set to 50 for memory/time efficiency on 10M rows
rf_model = RandomForestClassifier(
    n_estimators=50, 
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_smote, y_train_smote)
print("Random Forest Training Completed Successfully")
```

```python
y_pred_rf = rf_model.predict(X_test)
y_prob_rf = rf_model.predict_proba(X_test)[:, 1]

rf_train_accuracy = rf_model.score(X_train_smote, y_train_smote)
rf_test_accuracy = rf_model.score(X_test, y_test)

print(f"Training Accuracy : {rf_train_accuracy:.4f}")
print(f"Testing Accuracy  : {rf_test_accuracy:.4f}")
```

```python
joblib.dump(rf_model, "RandomForest_Model.pkl")
print("Random Forest Model Saved Successfully")
```

```python
### Random Forest Summary

• Random Forest was trained using the balanced training dataset.

• Ensemble learning enables the model to capture complex fraud patterns beyond linear relationships.

• Probability scores and class predictions were generated for the untouched testing dataset.

• The trained model has been saved for comparative evaluation alongside Logistic Regression.
```

```python
print("=" * 60)
print("Training XGBoost Model")
print("=" * 60)

xgb_model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=6,
    random_state=42,
    eval_metric="logloss",
    tree_method="hist", # Enterprise speed optimization for 10M+ rows
    n_jobs=-1
)

xgb_model.fit(X_train_smote, y_train_smote)
print("XGBoost Training Completed Successfully")
```

```python
y_pred_xgb = xgb_model.predict(X_test)
y_prob_xgb = xgb_model.predict_proba(X_test)[:, 1]

xgb_train_accuracy = xgb_model.score(X_train_smote, y_train_smote)
xgb_test_accuracy = xgb_model.score(X_test, y_test)

print(f"Training Accuracy : {xgb_train_accuracy:.4f}")
print(f"Testing Accuracy  : {xgb_test_accuracy:.4f}")
```

```python
joblib.dump(xgb_model, "XGBoost_Model.pkl")
print("XGBoost Model Saved Successfully")
```

```python
### XGBoost Summary

• XGBoost was trained using the balanced training dataset.

• Gradient boosting enables the model to capture highly complex fraud patterns while maintaining strong generalization.

• Probability scores and predictions were generated on the untouched testing dataset.

• The trained model has been saved for final evaluation and comparison with Logistic Regression and Random Forest.
```
